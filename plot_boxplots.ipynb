```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

## ssVERDICT with DNN

```python
f_ic_tumor = np.loadtxt('segmented_files/f_ic_tumor.csv', delimiter=',')
f_ic_normal = np.loadtxt('segmented_files/f_ic_normal.csv', delimiter=',')
```

```python
sns.boxplot(data=[f_ic_normal, f_ic_tumor], palette='Blues', width=0.25)
plt.xticks([0, 1], ['Normal', 'Tumor'])
plt.ylabel('f_ic')
plt.title('f_ic Distribution for Tumor and Normal Samples')
plt.grid(axis='y')
plt.show()
```

```python
from scipy.stats import wilcoxon
wilcoxon(f_ic_tumor, f_ic_normal)
```
